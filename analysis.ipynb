```python
# select the rows where email is not null
df_with_email = df_account[df_account['email'].notnull()]
print(f"Number of rows with email: {len(df_with_email)}")
```

```python
# select the email column which has @gwu.edu or @gwmail.gwu.edu or @email.gwu.edu
df_gwu_email = df_with_email[df_with_email['email'].str.contains('@gwu.edu|@gwmail.gwu.edu|@email.gwu.edu')]
print(f"Number of rows with gwu email: {len(df_gwu_email)}")
```
